```python
titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].mean())
titanic_df['Embarked'] = titanic_df['Embarked'].fillna(titanic_df['Embarked'].mode() [0])
titanic_df['Cabin'] = titanic_df['Cabin'].fillna('B96')
titanic_df.isnull().sum()
```

***To analyze the data perfectly we have to fill the values in the age , cabin and embarked. For age we used mean and for cabina and embarked we used mode***

```python
titanic_df.info()
```

```python
titanic_df.describe()
```

```python
def load_titanic_data(titanic_df):
  titanic_df['Survived'] = titanic_df['Survived'].replace({
      0: 'No Survived',
      1: 'Survived'
  })
  titanic_df['Pclass'] = titanic_df['Pclass'].replace({
      1: 'First Class',
      2: 'Second Class',
      3: 'Third Class'
  })
  titanic_df['Embarked'] = titanic_df['Embarked'].replace({
      'C': 'Cherbourg',
      'Q': 'Queenstown',
      'S': 'Southampton'
  })
  return titanic_df
titanic_df = load_titanic_data(titanic_df)
display(titanic_df)
```

***Before doing EDA we have to replace from values to analyzing the data more affectively. There are some volues in the Survived and Pclass which was written as 0,1,3 so we replace then with there correct values***

## ***what number of male and female was present in the ship ?***

```python
Male_VS_Female = titanic_df['Sex'].value_counts()
plt.bar(Male_VS_Female.index, Male_VS_Female.values)
plt.xlabel('Sex')
plt.ylabel('Count')
plt.title('Male VS Female')
plt.show()
```

***From the graph we can see that number of was greater then female***

## ***what was the common found in the in the titanic ?***

```python
sns.histplot(titanic_df['Age'], kde=True)
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Age Distribution')
plt.show()
```

***We can see the whose age was around 30 was common***

## ***how many people Survived the incident?***

```python
plt.bar(titanic_df['Survived'].value_counts().index, titanic_df['Survived'].value_counts().values)
plt.xlabel('Survived')
plt.ylabel('Count')
plt.title('Survived Distribution')
plt.show()
```

***We can see that out of 850 people which are in the given data set only 350 survived***

# ***which class was has the most people ?***

```python
plt.bar(titanic_df['Pclass'].value_counts().index, titanic_df['Pclass'].value_counts().values)
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.title('Pclass Distribution')
plt.show()
```

***Most people was in third class***

# ***Did females survive more than males?***

```python
Survivel_count_by_sex = titanic_df.groupby(['Sex', 'Survived']).size().unstack()
Survivel_count_by_sex.plot(kind='bar', stacked=True)
plt.xlabel('Sex')
plt.ylabel('Count')
plt.title('Survived Count by Sex')
plt.show()
```

***Female survival rate was higher.***

# ***Which age people survival the most ?***

```python
sns.histplot(data=titanic_df,x='Age',hue='Survived',bins=20,multiple='stack')
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Survival Count by Age')
plt.show()
```

***the people of age 30 has the highest rate of survival***

## ***which class has the most survived rate ?***

```python
survival_rate_by_class = titanic_df.groupby(['Pclass', 'Survived']).size().unstack()
survival_rate_by_class.plot(kind='bar', stacked=True)
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.title('Survival Count by Pclass')
plt.show()
```

***First class has the most survival rate***

# ***which Embarked has the most survived rate ?***

```python
survival_count_by_embarked = titanic_df.groupby(['Embarked', 'Survived']).size().unstack()
survival_count_by_embarked.plot(kind='bar', stacked=True)
plt.xlabel('Embarked')
plt.ylabel('Count')
plt.title('Survival Count by Embarked')
plt.xticks(rotation=0)
plt.show()
```

***Southampton have the most survival***

# ***Hypothesis 1: Females had a higher survival rate than males.***

```python
sns.countplot(x='Sex', hue='Survived', data=titanic_df)
plt.xlabel('Sex')
plt.ylabel('Count')
plt.title('Survival Count by Sex')
plt.show()
```

***Reason: During the Titanic disaster, women and children were often given priority during evacuation.***

***Result: The analysis showed that female passengers had a significantly higher survival rate compared to male passengers.***

***Conclusion: The hypothesis is supported by the data.***

### ***Hypothesis 2: First-class passengers were more likely to survive than third-class passengers.***

```python
sns.countplot(x='Pclass', hue='Survived', data=titanic_df)
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.title('Survival Count by Pclass')
plt.show()
```

***Reason: First-class passengers had better access to lifeboats and evacuation routes.***

***Result: First-class passengers showed a higher survival rate than passengers in second and third class.***

***Conclusion: The hypothesis is supported by the data.***

### ***Hypothesis 3: Younger passengers had a higher chance of survival.***

```python
titanic_df['AgeGroup'] = pd.cut(
    titanic_df['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'])
sns.countplot(
    data=titanic_df,
    x='AgeGroup',
    hue='Survived')
plt.title('Survival Count by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Passengers')
plt.show()
```

***Reason: Children were often prioritized during rescue operations.***

***Result: The analysis indicated that younger passengers generally had better survival outcomes than many adult passengers.***

***Conclusion: The hypothesis is partially supported by the data.***
