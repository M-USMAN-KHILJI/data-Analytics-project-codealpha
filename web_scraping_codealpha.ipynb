```python
pakwheels_df = pd.read_csv('dataset_pakwheels-scraper_2026-06-03_14-44-46-948.csv')
display(pakwheels_df)
```

```python
pakwheels_df.isnull().sum()
```

***looking for missing values before analyzing***

```python
pakwheels_df['body_type'] = pakwheels_df['body_type'].fillna(pakwheels_df['body_type'].mode()[0])
pakwheels_df['thumbnail'] = pakwheels_df['thumbnail'].fillna(pakwheels_df['thumbnail'].mode()[0])
pakwheels_df.isnull().sum()
```

***Filling the missing values by taking mode to analyze data more effectively***

```python
pakwheels_df.info()
```

```python
pakwheels_df.describe()
```

## **What car types are commonly fount in the pakwheels dataset ?**

```python
body_type_counts = pakwheels_df['body_type'].value_counts()
plt.figure(figsize=(10, 6))
sns.barplot(x=body_type_counts.index, y=body_type_counts.values)
plt.xlabel('Body Type')
plt.ylabel('Count')
plt.title('Body Type Distribution')
plt.show()
```

***Sedans are the most common cars in the pakwheels dataset***

## ***which cities has the most cars to for sale ?***

```python
plt.figure(figsize=(10, 6))
sns.histplot(pakwheels_df['city'].head(50), kde=True)
plt.xlabel('City')
plt.ylabel('Count')
plt.title('City Distribution')
plt.xticks(rotation=45)
plt.show()
```

***according the dataset lahore has the most for sale***

## ***which companies the most cars ?***

```python
company_counts = pakwheels_df['make'].value_counts()
plt.figure(figsize=(10, 6))
sns.barplot(x=company_counts.index, y=company_counts.values)
plt.xlabel('Company')
plt.ylabel('Count')
plt.title('Company Distribution')
plt.show()
```

***we can see that toyota is dominating the market***

***Which car model is the most common ?***

```python
plt.figure(figsize=(10, 6))
sns.histplot(pakwheels_df['model'], kde=True)
plt.xlabel('Model')
plt.ylabel('Count')
plt.title('Model Distribution')
plt.xticks(rotation=90)
plt.show()
```

***Corolla is the used car***

## ***Most common car for sale in the pakwheels dataset ?***

```python
plt.figure(figsize=(10, 6))
sns.histplot(pakwheels_df['year'], kde=True)
plt.xlabel('Year')
plt.ylabel('Count')
plt.title('Year Distribution')
plt.show()
```

***2020 car are the most common***

## ***which engine cc is most common ?***

```python
engine_cc_counts = pakwheels_df['engine_cc'].value_counts()
plt.figure(figsize=(10, 6))
sns.barplot(x=engine_cc_counts.index, y=engine_cc_counts.values)
plt.xlabel('Engine CC')
plt.ylabel('Count')
plt.title('Engine CC Distribution')
plt.show()
```

***1300 cc is the most famous engine***

# ***which price range is commom ?***

```python
price_counts = pakwheels_df['price'].head(50).value_counts()
plt.figure(figsize=(10, 6))
sns.barplot(x=price_counts.index, y=price_counts.values)
plt.xlabel('Price')
plt.ylabel('Count')
plt.title('Price Distribution')
plt.xticks(rotation=45)
plt.show()
```

***50 lakh is the common price range***

```python
sns.histplot(pakwheels_df['color'].head(20) , kde=True)
plt.xlabel('color')
plt.ylabel('Count')
plt.title('color Distribution')
plt.xticks(rotation=45)
plt.show()
```

# ***which city has most common car model ?***

```python
city_model = pakwheels_df.groupby(['city', 'model']).size().head(30).reset_index(name='count')
plt.figure(figsize=(12,6))
sns.barplot(data=city_model, x='city', y='count', hue='model')
plt.xticks(rotation=45)
plt.title('Number of Cars by City and Model')
plt.show()
```

# ***what is price of the cars ?***

```python
price_model = pakwheels_df.groupby(['price' , 'model']).size().head(25).reset_index(name='count')
plt.figure(figsize=(12,6))
sns.barplot(data=price_model, x='price', y='count', hue='model')
plt.xlabel('Price')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.title('Number of Cars by Price and Model')
plt.show()
```

***city has the highest values***

## ***what is the price of engine cc and price ?***

```python
engine_cc_price = pakwheels_df.groupby(['engine_cc', 'price']).size().head(25).reset_index(name='count')
plt.figure(figsize=(12,6))
sns.boxplot(data=pakwheels_df,x='engine_cc',y='price')
plt.xlabel('Engine CC')
plt.ylabel('Price')
plt.title('Engine CC vs Price')
plt.xticks(rotation=45)
plt.show()
```

***1300 cc is have most price***

***which city has the most price ?***

```python
city_price = pakwheels_df.groupby(['city', 'price']).size().head(20).reset_index(name='count')
plt.figure(figsize=(12,6))
sns.barplot(data=city_price, x='city', y='count', hue='price')
plt.xlabel('City')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.title('Number of Cars by City and Price')
plt.show()
```
