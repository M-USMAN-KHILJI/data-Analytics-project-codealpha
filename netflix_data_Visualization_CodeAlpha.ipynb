```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
netflix_df = pd.read_csv('netflix_titles.csv')
display(netflix_df)
```

```python
netflix_df.info()
```

```python
netflix_df.isnull().sum()
```

```python
netflix_df['date_added'] = netflix_df['date_added'].fillna(netflix_df['date_added'].mode()[0])
netflix_df['country'] = netflix_df['country'].fillna(netflix_df['country'].mode()[0])
netflix_df['cast'] = netflix_df['cast'].fillna(netflix_df['cast'].mode()[0])
netflix_df['director'] = netflix_df['director'].fillna(netflix_df['director'].mode()[0])
netflix_df['rating'] = netflix_df['rating'].fillna(netflix_df['rating'].mode()[0])
netflix_df['duration'] = netflix_df['duration'].fillna(netflix_df['duration'].mode()[0])
netflix_df.isnull().sum()
```

```python
display(netflix_df)
```

```python
netflix_df['month'] = pd.to_datetime(netflix_df['date_added'] , format='mixed').dt.month.astype(int)
display(netflix_df)
```

```python
def load_netflix_data(netflix_df):
  netflix_df['month'] = netflix_df['month'].replace({
      1: 'January',
      2: 'February',
      3: 'March',
      4: 'April',
      5: 'May',
      6: 'June',
      7: 'July',
      8: 'August',
      9: 'September',
      10: 'October',
      11: 'November',
      12: 'December'
  })
  return netflix_df
netflix_df = load_netflix_data(netflix_df)
display(netflix_df)
```

```python
type = netflix_df['type'].value_counts()
plt.bar(type.index , type.values)
plt.xlabel('Type')
plt.ylabel('Count')
plt.title('Distribution of Types')
plt.show()
```

***Netflix initially built its platform around movies, which still dominate the catalog.***

```python
sns.histplot(netflix_df['release_year'] , kde=True , bins = 50)
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.title('Distribution of Release Years')
plt.show()
```

***Netflix experienced major growth after becoming a global streaming service, especially in the late 2010s.***

```python
plt.bar(netflix_df['month'].value_counts().index , netflix_df['month'].value_counts().values)
plt.xlabel('month')
plt.ylabel('count')
plt.title('Distribution of months')
plt.xticks(rotation=45)
plt.show()
```

***his may reflect Netflix’s mid-year release strategy to attract viewers during summer holidays.***

```python
sns.histplot(netflix_df['rating'] , kde=True , bins = 50)
plt.xlabel('rating')
plt.ylabel('count')
plt.title('Distribution of rating')
plt.xticks(rotation=45)
plt.show()
```

***Netflix prioritizes adult content, which aligns with its global audience base.***

```python
plt.figure(figsize=(10,10))
plt.pie(netflix_df['rating'].value_counts() , labels=netflix_df['rating'].value_counts().index , autopct='%1.1f%%')
plt.title('Distribution of rating')
plt.show()
```

```python
top_genres = netflix_df['listed_in'].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title('Top 10 Genres on Netflix')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.show()
```

```python
types_Vs_rating = netflix_df.groupby(['type' , 'rating']).size().reset_index(name='count')
plt.figure(figsize=(10,10))
sns.barplot(x='type' , y='count' , hue='rating' , data=types_Vs_rating)
plt.xlabel('Type')
plt.ylabel('Count')
plt.title('Distribution of Types vs Rating')
plt.xticks(rotation=45)
plt.show()
```

***In movies 2000+ movies has TV-MG ratings and in TV-Show more then 1000+ TV-show has TV-MG rating***

```python
top_countries = netflix_df['country'].value_counts().head(25).index
country_type = netflix_df[netflix_df['country'].isin(top_countries)].groupby(['country', 'type']).size().unstack(fill_value=0)
country_type.plot(kind='bar',stacked=True,figsize=(12,6))
plt.title('Movies and TV Shows by Country')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

***The United States dominates Netflix's content library, contributing significantly more titles than any other country. This suggests that Netflix has a strong focus on U.S.-produced content and may prioritize English-speaking audiences.***

```python
type_year = netflix_df.groupby(['release_year', 'type']).size().reset_index(name='count')

plt.figure(figsize=(12,6))

sns.lineplot(data=type_year,x='release_year',y='count',hue='type')

plt.title('Netflix Content Released by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

***Netflix experienced major growth after becoming a global streaming service, especially in the late 2010s.***
